```python
# Imports
import pandas as pd
pd.options.display.precision = 2
pd.options.display.max_rows = 20

import numpy as np 

# Constants
DATA_DIR = "../../data_t1_stats_for_data_scientists/"
SEPARATOR = "\n------------------------------------------------------------\n"
```

```python
# Read the dataset
dataset = "face-data.csv"
df = pd.read_csv(DATA_DIR+dataset)
```

```python
# Summarize Data
# Print first and last points of measurement.
print("HEAD: ")
print(df.head(5))
print(SEPARATOR)
print("TAIL: ")
print(df.tail(5))
print(SEPARATOR)
# Print all the descriptive statistics
print("Descriptive Stats")
print(df.describe(include='all'))
print(SEPARATOR)
```

```python
# Show descriptive stats for only categorical data
df.describe(include=[object])
```

```python
# Show frequencies of various categorical data
# Double bracket in action.

df = df[df['gender'].isin(['Male', 'Female'])]

for column in ['gender', 'age', 'edu']:
  print(df[column].value_counts(sort=True, ascending=True))
  print(SEPARATOR)
```
